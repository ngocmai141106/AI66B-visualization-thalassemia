```python
import pandas as pd
import numpy as np
from pathlib import Path

#rmb to change paths if there's any change in file structure
GBD_PATH = Path("../data/raw/GBD_raw.csv")
WDI_PATH = Path("../data/raw/WDI_raw.csv")
#load dts
gbd = pd.read_csv(GBD_PATH)
wdi = pd.read_csv(WDI_PATH)

print("GBD dataset loaded successfully!")
print("WDI dataset loaded successfully!")
```

```python
#GBD dataset inspection
print("GBD Shape:", gbd.shape)
display(gbd.head())
print("\nGBD Columns:")
print(gbd.columns.tolist())

#WDI dataset inspection
print("WDI Shape:", wdi.shape)
display(wdi.head())
print("\nWDI Columns:")
print(wdi.columns.tolist())
```

```python
#indentify gbd variables
gbd_variables = [
    "location_name",
    "year",
    "cause_name",
    "measure_name",
    "metric_name",
    "age_name",
    "sex_name",
    "val"
]
display(gbd[gbd_variables].head())

#identify wdi variables/indicators
wdi_indicators = (
    wdi[["Series Name", "Series Code"]]
    .dropna()
    .drop_duplicates()
)
display(wdi_indicators)
```

```python
#country and years problem inspection
#GBD
print("GBD Countries:")
print(sorted(gbd["location_name"].unique()))
print("\nGBD Years:")
print(sorted(gbd["year"].unique()))
print("\nNumber of countries:", gbd["location_name"].nunique())
print("Number of years:", gbd["year"].nunique())

#WDI
#remove metadata/footer rows at the end of the csv file
wdi_data = wdi[wdi["Series Code"].notna()].copy()
print("WDI Countries:")
print(sorted(wdi_data["Country Name"].unique()))
year_columns = [
    col for col in wdi.columns
    if "[YR" in col
]
print("\nWDI Years:")
print(year_columns)
```

```python
#check for missing values
print("GBD Missing Values:")
display(gbd.isnull().sum())
print("WDI Missing Values:")
display(wdi_data.isnull().sum())

# Check duplicates
print("GBD duplicate rows:", gbd.duplicated(["location_name", "year"]).sum())
print("WDI duplicate country-indicator rows:", wdi_data.duplicated(["Country Code", "Series Code"]).sum())

#all problem handlings will be in the next notebook, if needed
```
